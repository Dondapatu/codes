```python

def Dec(func):
    def Inner(x,y):
        print("starting this function")
        func(x*y,x/y)
    print(f"Func:{func}")
    print(f"Inner:{Inner}")
    return Inner

@Dec
def fun(a,b):
    print(a,b)
    print(a+b)
fun(10,20)
```

```python
def Dec(func):
    def Inner(x, y):
        if isinstance(x,int) and isinstance (y,int):
            print("sending integer")
            func(x,y)
        elif isinstance(x,str) and isinstance (y,str):
            print("sending string")
            func(x,y)
        else:
            print("Invalid argument")
    return Inner
@Dec
def fun(a, b):
    print(a, b)
    print(a + b)

fun(10,20)


```
